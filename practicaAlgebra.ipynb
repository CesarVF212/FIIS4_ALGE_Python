### Problema 1

Apartado A) Resuelve AX + 3B = -5X

$$
AX + 5X = -3B \\
(A + 5)X = -3B \\
(A + 5)^{-1} * (A + 5)X = (A + 5)^{-1} * (-3B) \\
X = (A + 5)^{-1} * (-3B) \\
$$

```python
A = np.array([[-6, -3], [0, -3]])
B = np.array([[-1, 0], [4, -2]])

I = np.eye(2) # Matriz identidad 2x2

resultado = np.linalg.inv(A + 5*I) @ (-3*B) 
print(resultado)
```

Apartado B) Resuelve ACX + 3B = 10I - X

$$
ACX + X = 10I - 3B \\
(AC + I)X = 10I - 3B \\
(AC + I)^{-1} * (AC + I)X = (AC + I)^{-1} * (10I - 3B) \\
X = (AC + I)^{-1} * (10I - 3B) \\
$$

```python
A = np.array([[1, 2], [0, 2]])
B = np.array([[2, -1], [-1, 5]])
C = np.array([[2, 6], [-1, -0.5]])
I = np.eye(2) # Matriz identidad 2x2

resultado = np.linalg.inv((A @ C) + I) @ (10*I - 3*B)
print(resultado)
```

### Problema 2

Dado el siguiente sistema de ecuaciones lineales para los distintos valores reales del
parámetro “a”. Implementa la discusión e implementación del sistema.

$$
\begin{cases}
    x  +  y - z  = 1 \\
    3x + ay + az = 5 \\
    4x + ay      = 5 \\
\end{cases}

$$

```python
import sympy as sp

a, x, y, z = sp.symbols('a x y z')
M = sp.Matrix([[1, 1, -1], [3, a, a], [4, a, 0]])
N = sp.Matrix([1, 5, 5])

# Resolvemos el determinante y guardamos los valores que lo hacen 0.
determinante = M.det()
valores_a = sp.solve(determinante, a)

# Estudiamos los casos. Todos los valores que no hagan 0 el determinante son compatibles. El resto los estudiamos a parte.
print("--- ESTUDIO DE CASOS ---")
print(f"\nCaso 1 (SCD). a ∉ {valores_a}\nEn este caso el determinante de M es distinto de 0, así que tiene rango 3. Existe una solución única para el sistema.")
solucion_scd = M.solve(N)

# Iteramos por todos los valores que hacen 0 el determinante para analizar su "comportamiento".
for i, val in enumerate(valores_a, start=2):

    print(f"\nCaso {i}. a = {val}")
    A_val = M.subs(a, val)
    Ab_val = A_val.row_join(N)
    rango_a = A_val.rank()
    rango_aumentada = Ab_val.rank()
    print(f"Rango(A) = {rango_a}, rango([A|A*]) = {rango_aumentada}")

    if rango_a == rango_aumentada:

        ecuaciones = list(A_val * sp.Matrix([x, y, z]) - N)

        if rango_a < 3:
            print("Sistema Compatible Indeterminado (SCI)")
        else:
            print("Sistema Compatible Determinado (SCD)")
            
    else:
        print("Sistema Incompatible (SI)")
```

### Problema 3

Dados los puntos P = (1, 2, -1), Q = (1, 3, 2) y la recta r: 2x + y - z = 6, x - y + z = 1.

Apartado a) Calcula la proyección ortogonal de P sobre r

Apartado b) Calcula la proyección ortogonal de PQ sobre el vector director de r

```python
P = np.array([1, 2, -1])
Q = np.array([1, 3, 2])
vec_PQ = Q - P

# r es una intersección de dos planos así que resolvemos el sistema para obtener vector director
n1 = np.array([2, 1, -1])
n2 = np.array([1, -1, 1])
vector_d = np.cross(n1, n2) # Vector director de r

# Sacamos un punto de r haciendo z = 0
# 2x + y = 6, x - y = 1 -> 3x = 7 -> x = 7/3, y = 4/3
punto_r = np.array([7/3, 4/3, 0])

# Vector RP
RP = P - punto_r

# Proyección de RP sobre r
proy_RP_vr = (np.dot(RP, vector_d) / np.dot(vector_d, vector_d)) * vector_d
P_proy = punto_r + proy_RP_vr

# Proyección de PQ sobre el vector director
proy_PQ_vr = (np.dot(vec_PQ, vector_d) / np.dot(vector_d, vector_d)) * vector_d

print(f"Proyección de P sobre r: {P_proy}")
print(f"Proyección de PQ sobre el director: {proy_PQ_vr}")
```
