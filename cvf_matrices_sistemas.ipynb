### _Problema 1 (2,5 puntos cada apartado)_

Implementar en Python el planteamiento y resolución de las siguientes ecuaciones matriciales.


a )
$$
AX + 3B = -5X \qquad 

A = \begin{pmatrix}
-6 & -3 \\
0 & -3
\end{pmatrix} \quad 

B = \begin{pmatrix}
-1 & -0 \\
4 & -2
\end{pmatrix}

$$

```python
# Declaramos las matrices:

A = np.array([[-6, 3], [0, -3]], dtype=int)
B = np.array([[-1, 0], [4, -2]], dtype=int)
I = np.eye(2)
```

Se despeja para X y luego resuelve con Python.

$$
X = \left( A - 5I \right)^{-1} * (- 3B)
$$

```python
inv = np.linalg.inv(A - 5 * I)
X_a = np.dot(inv, (-3 * B))
```

```python
print(X_a)
```

b )
$$
ACX + 3B = -10I - X \qquad 

A = \begin{pmatrix}
1 & 2 \\
0 & 2  
\end{pmatrix} \quad 

B = \begin{pmatrix}
2 & -1 \\
-1 & 5
\end{pmatrix}\quad 

C = \begin{pmatrix}
2 & 6 \\
-1 & -0.5
\end{pmatrix}

$$

```python
A = np.array([[1, 2], [0, 2]], dtype=int)
B = np.array([[2, -1], [-1, 5]], dtype=int)
C = np.array([[2, 6], [-1, -0.5]], dtype=float)
I = np.eye(2)
```

Otra vez, despejamos el sistema:
$$
X = (AC + I)^{-1} * \left( 10 I - 3B\right)
$$

```python
left = np.linalg.inv(np.dot(A, C) + I)
right = 10 * I - 3 * B

X_b = np.dot(left, right)
```

```python
print(X_b)
```

### Problema 2 (5 puntos)
Dado el siguiente sistema de ecuaciones lineales para los distintos valores reales de los parámetros a y b.

$$
\begin{cases}
 ax + y + bz = 0        \\
 x + ay + bz = b - 2    \\
 x + ay + bz = a - 1 
\end{cases}
$$

Para los casos:
- b = 1, a = 2
- b = 2, a = 3
- a = 1, b = 2

```python
# Declaramos un doble array con las posibilidades 
a = [2, 3, 1] 
b = [1, 2, 2]


# Declaramos las matrices.
for i in range(0, len(a)) :
    
    a_i = a[i]
    b_i = b[i]
    print(f"Sistema con parámetros [a = {a_i}, b = {b_i}]")

    A = np.array([[a_i, 1, b_i], [1, a_i, b_i], [1, a_i, b_i]], dtype=int)
    B = np.array([[0], [b_i - 2], [a_i - 1]], dtype=int)

    # Concatenamos la matriz A con su ampliada.
    A_added = np.concatenate((A, B), axis=1)

    # Vemos si los rangos coinciden
    rankA = np.linalg.matrix_rank(A)
    rankA_added = np.linalg.matrix_rank(A_added)

    if(rankA == rankA_added):
        print("Los sistemas son incompatibles")

    else:


        # Obtenemos el número de filas y de columnas.
        num_rows, num_cols = A.shape
        if(rankA != num_rows):
            print("Los sitemas son compatibles indeterminados")
        
        else:
            print("Los sistemas son compatibles.")



```
