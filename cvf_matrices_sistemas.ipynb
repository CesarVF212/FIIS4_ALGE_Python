### _Problema 1 (2,5 puntos cada apartado)_

Implementar en Python el planteamiento y resolución de las siguientes ecuaciones matriciales.


a )
$$
AX + 3B = -5X \qquad 

A = \begin{pmatrix}
-6 & -3 \\
0 & -3
\end{pmatrix} \quad 

B = \begin{pmatrix}
-1 & 0 \\
4 & -2
\end{pmatrix}

$$

```python
# Declaramos las matrices:

A = np.array([[-6, 3], [0, -3]], dtype=int)
B = np.array([[-1, 0], [4, -2]], dtype=int)
I = np.eye(2)
```

Se despeja para X y luego resuelve con Python.

$$
X = \left( A - 5I \right)^{-1} * (- 3B)
$$

```python
inv = np.linalg.inv(A - 5 * I)
X_a = np.dot(inv, (-3 * B))
```

```python
print(X_a)
```

b )
$$
ACX + 3B = -10I - X \qquad 

A = \begin{pmatrix}
1 & 2 \\
0 & 2  
\end{pmatrix} \quad 

B = \begin{pmatrix}
2 & -1 \\
-1 & 5
\end{pmatrix}\quad 

C = \begin{pmatrix}
2 & 6 \\
-1 & -0.5
\end{pmatrix}

$$

```python
A = np.array([[1, 2], [0, 2]], dtype=int)
B = np.array([[2, -1], [-1, 5]], dtype=int)
C = np.array([[2, 6], [-1, -0.5]], dtype=float)
I = np.eye(2)
```

Otra vez, despejamos el sistema:
$$
X = (AC + I)^{-1} * \left( 10 I - 3B\right)
$$

```python
left = np.linalg.inv(np.dot(A, C) + I)
right = 10 * I - 3 * B

X_b = np.dot(left, right)
```

```python
print(X_b)
```

### Problema 2 (5 puntos)
Dado el siguiente sistema de ecuaciones lineales para los distintos valores reales de los parámetros a y b.

$$
\begin{cases}
 x + y - z &= 0        \\
 3x + ay + az &= 5    \\
 4x + ay &= 5 
\end{cases}
$$

```python
# Variable de sympy para nuestro parámetro.
a = symbols('a')

# Definición del sistema
A = Matrix([
    [1,  1, -1],
    [3,  a,  a],
    [4,  a,  0]
])

# Matriz de coeficientes (columna).
b = Matrix([1, 5, 5])
```

```python
def rouche_frobenius(A, b):

    # Creamos la matriz ampliada.
    Ab = A.row_join(b)

    # Extraemos rango y número de variables.
    n_vars   = A.shape[1]
    rango_A  = A.rank()
    rango_Ab = Ab.rank()

    # Resolvemos el determinantes, obteniendo los valores críticos.
    det_A = A.det()
    valores_criticos = solve(det_A, a)
    print(f"det(A) = {factor(det_A)}")
    print(f"Valores críticos de a: {valores_criticos}\n")

    # --- CASO GENERAL --- #
    print("=== CASO GENERAL ===")
    print(f"  Rango de A:     {rango_A}")
    print(f"  Rango de [A|b]: {rango_Ab}")
    print(f"  Nº incógnitas:  {n_vars}")

    if rango_A == rango_Ab == n_vars:
        sol = A.solve(b)
        print("\tSistema COMPATIBLE DETERMINADO (solución única):")
        for i, val in enumerate(sol):
            print(f"\tx{i+1} = {simplify(val)}")

    # --- CASOS CRÍTICOS --- ##

    for val in valores_criticos:

        print(f"\n=== CASO a = {val} ===")
        
        A_s  = A.subs(a, val)
        Ab_s = Ab.subs(a, val)
        r_A  = A_s.rank()
        r_Ab = Ab_s.rank()

        print(f"  Rango de A:     {r_A}")
        print(f"  Rango de [A|b]: {r_Ab}")

        # Caso incompatible.
        if r_A != r_Ab:
            print("\tSistema INCOMPATIBLE (sin solución)")

        # Caso compatible indeterminado.
        elif r_A == r_Ab != n_vars:
            grados = n_vars - r_A
            print(f"\tSistema COMPATIBLE INDETERMINADO ({grados} grado(s) de libertad)")
            


rouche_frobenius(A, b)
```
